# NusaMart Sales Analysis

Bounded fallback notebook for BelajarData Module 04. Load the predefined data, inspect it, calculate revenue, merge the product master, and compare September performance.

```python
from io import StringIO
import pandas as pd

transactions_csv = """order_id,order_date,region,channel,product_id,quantity,unit_price
O-1001,2025-08-03,West,Online,P-1001,2,120
O-1002,2025-08-08,East,Store,P-2001,1,120
O-1003,2025-08-17,West,Online,P-3001,3,30
O-1004,2025-09-02,West,Store,P-1002,1,80
O-1005,2025-09-06,East,Online,P-1001,1,120
O-1006,2025-09-11,West,Online,P-2001,2,90
O-1007,2025-09-15,South,Store,P-3001,4,25
O-1008,2025-09-21,West,Store,P-1001,1,120
"""
products_csv = """product_id,category,brand
P-1001,Electronics,NusaSound
P-1002,Electronics,NusaHome
P-2001,Home,NusaHome
P-3001,Grocery,NusaFresh
"""

transactions = pd.read_csv(StringIO(transactions_csv), parse_dates=["order_date"])
products = pd.read_csv(StringIO(products_csv))
transactions["revenue"] = transactions["quantity"] * transactions["unit_price"]
transactions["month"] = transactions["order_date"].dt.to_period("M").astype(str)
sales = transactions.merge(products, on="product_id", how="left")
september = sales[sales["month"] == "2025-09"]
summary = september.groupby("category", as_index=False)["revenue"].sum().sort_values("revenue", ascending=False)
display(summary)
print("September revenue:", september["revenue"].sum())
```
